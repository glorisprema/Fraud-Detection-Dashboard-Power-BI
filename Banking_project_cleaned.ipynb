```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

np.random.seed(42)
n=100000
start_date = datetime(2023,1,1)
dates = [start_date + timedelta(days=int(np.random.randint(0,650))) for _ in range(n)]

df = pd.DataFrame({
    'Transaction_ID': [f'TXN{1000000+i}' for i in range(n)],
    'Date': dates,
    'Customer_ID': [f'CUST{int(np.random.randint(1000,9999))}' for _ in range(n)],
    'Customer_Name': np.random.choice(['John Smith',' Priya Patel ','AHMED ALI','emily johnson','  David Lee'], n),
    'Age': np.random.randint(18,75,n).astype(object),
    'Account_Type': np.random.choice(['Savings','savings','SAVINGS','Checking','CHECKING','Credit','Loan'], n),
    'Transaction_Type': np.random.choice(['Deposit','deposits','Withdrawal','WITHDRAWAL','Transfer','Payment'], n),
    'Amount': np.round(np.random.exponential(500, n) + 10, 2).astype(object),
    'Branch': np.random.choice(['Hamilton Downtown',' hamilton downtown ','Toronto Main','TORONTO MAIN','Oakville', None], n, p=[0.3,0.1,0.3,0.05,0.2,0.05]),
    'Channel': np.random.choice(['Online','online','ATM','In-Branch',' Mobile App '], n),
    'Balance_After': np.round(np.random.uniform(100, 50000, n), 2),
    'Is_Fraud': np.random.choice([0,1,'Yes','No'], n, p=[0.49,0.01,0.25,0.25])
})

# REAL MESS
df.loc[np.random.choice(n, 4500, False), 'Amount'] = np.nan
df.loc[np.random.choice(n, 3200, False), 'Age'] = np.nan
df.loc[np.random.choice(n, 800, False), 'Amount'] = -500
df.loc[np.random.choice(n, 300, False), 'Age'] = 200
df.loc[np.random.choice(n, 50, False), 'Amount'] = 150000
df.loc[np.random.choice(n, 700, False), 'Date'] = '2023-13-45'
dupes = df.sample(800)
df = pd.concat([df, dupes], ignore_index=True)

df.to_csv('project2_REAL_MESSY_100k.csv', index=False)
print("Messy file created:", len(df))
print(df.isnull().sum())
```

```python
# STEP 1: Check Age - in simple numbers
print("--- AGE CHECK ---")
Q1_age = df['Age'].quantile(0.25)
Q3_age = df['Age'].quantile(0.75)
IQR_age = Q3_age - Q1_age

print(f"Q1 (25%): {Q1_age}")
print(f"Q3 (75%): {Q3_age}")
print(f"IQR (middle spread): {IQR_age}")

# STEP 2: Calculate normal range
lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age

print(f"\nNormal Age should be between {lower_age:.1f} and {upper_age:.1f}")

# STEP 3: Find weird ages
weird_age = df[(df['Age'] < lower_age) | (df['Age'] > upper_age)]
print(f"\nFound {len(weird_age)} weird Age rows")
print(weird_age[['Age', 'Branch', 'Amount']].head(10))

# STEP 4: Same for Amount - to catch fraud
print("\n--- AMOUNT CHECK (Fraud Check) ---")
Q1_amt = df['Amount'].quantile(0.25)
Q3_amt = df['Amount'].quantile(0.75)
IQR_amt = Q3_amt - Q1_amt

lower_amt = Q1_amt - 1.5 * IQR_amt
upper_amt = Q3_amt + 1.5 * IQR_amt

print(f"Normal Amount should be between {lower_amt:.1f} and {upper_amt:.1f}")

weird_amount = df[df['Amount'] > upper_amt]  # only high amount is fraud
print(f"\nFound {len(weird_amount)} high Amount (possible fraud) rows")
print(weird_amount[['Age', 'Branch', 'Amount', 'Transaction_Type']].head(10))
```

```python
q_velocity_final = pd.read_sql("""
SELECT 
  d.Customer_ID,
  d.Date as Deposit_Time,
  w.Date as Withdrawal_Time,
  CAST((JULIANDAY(w.Date) - JULIANDAY(d.Date)) as INTEGER) as Days_Diff,
  d.Amount as Deposit_Amt,
  w.Amount as Withdrawal_Amt
FROM banking_transaction d
JOIN banking_transaction w 
  ON d.Customer_ID = w.Customer_ID
  AND w.Date > d.Date
WHERE d.Transaction_Type = 'Deposit'
  AND w.Transaction_Type = 'Withdrawal'
  AND (JULIANDAY(w.Date) - JULIANDAY(d.Date)) <= 7
ORDER BY Days_Diff ASC
LIMIT 20
""", conn)
```

```python
print(q_velocity_final)
```

```python
q_rule = pd.read_sql("""
SELECT 
  Channel,
  Transaction_Type,
  COUNT(*) as Flagged_Txn,
  SUM(Amount) as Blocked_Amount,
  ROUND(AVG(Amount),2) as Avg_Fraud_Amount
FROM banking_transaction
WHERE Is_Fraud = 1
  AND Channel = 'Online'
  AND Amount > 500
GROUP BY Channel, Transaction_Type
""", conn)

print(q_rule)
```

```python
Risk_Measure=pd.read_sql("""SELECT 
                        case when Amount<500 THEN 'small'
                        when Amount Between 500 and 5000 then 'Medium'
                        Else 'Large(>5K)-Flag'
                        End as risk_bucket,
                         count(*)as tx_count,
                         sum(case when is_fraud=1 then Amount else 0 End)as fraud_amount
                         From Banking_transaction
                         Group by risk_bucket
                         order by fraud_amount Desc""",conn)
```

```python
print(Risk_Measure)
```

```python
Risk_flags=pd.read_sql("""SELECT
  t.*,
  -- q3
  CASE WHEN Channel='Online' AND Is_Fraud=1 THEN 1 ELSE 0 END as is_online_fraud,
  -- q4_fraud
  CASE WHEN Transaction_Type='Withdrawal' AND Is_Fraud=1 THEN 1 ELSE 0 END as is_withdrawal_fraud,
  -- q4_channel_fraud (your $2.3M top risk)
  CASE WHEN Channel='Online' AND Transaction_Type='Withdrawal' AND Is_Fraud=1 THEN 1 ELSE 0 END as is_online_withdrawal_fraud,
  CASE WHEN Channel='Online' AND Transaction_Type='Deposit' AND Is_Fraud=1 THEN 1 ELSE 0 END as is_online_deposit_fraud,
  -- q_rule (3,684 = $4.7M)
  CASE WHEN Channel='Online' AND Amount>500 AND Is_Fraud=1 THEN 1 ELSE 0 END as q_rule_flag,
  -- Risk_Measure (Large 54)
  CASE WHEN Amount<500 THEN 'small' WHEN Amount<=5000 THEN 'Medium' ELSE 'Large(>5K)-Flag' END as risk_bucket
FROM banking_transaction t""",conn)

```
